```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. 데이터 입력 (관측 빈도)
# 완구 A, B, C, D를 선택한 어린이 수
observed = np.array([24, 28, 38, 30])

# 2. 기대 빈도 설정 (총 120명을 4로 나눈 값 = 각각 30명씩 균등하다고 가정)
expected = np.array([30, 30, 30, 30])

# 3. 적합도 검정 수행
chisq_stat, p_value = stats.chisquare(f_obs=observed, f_exp=expected)

# 4. 결과 출력
print("=== 일원 카이제곱 검정 (적합도 검정) 결과 ===")
print(f"카이제곱 통계량 (Chi-square statistic): {chisq_stat:.4f}")
print(f"p-값 (p-value): {p_value:.4f}")

# 5. 가설 검정 결론
alpha = 0.05
if p_value < alpha:
    print(f"결론: p-value가 {alpha}보다 작으므로 귀무가설을 기각합니다.")
    print("-> 완구 종류에 따른 선호도에는 통계적으로 유의미한 차이가 존재합니다.")
else:
    print(f"결론: p-value가 {alpha}보다 크므로 귀무가설을 채택합니다.")
    print("-> 완구 종류에 따른 선호도 차이가 있다고 볼 수 없습니다.")
```

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. 분할표(Contingency Table) 생성
# 행: 주간근무, 야간근무 / 열: 정상 제품 수, 불량 제품 수
data = np.array([
    [185, 15],  # 주간 근무 (정상 185, 불량 15)
    [170, 30]   # 야간 근무 (정상 170, 불량 30)
])

# 시각적으로 확인하기 위해 DataFrame으로 변환 (선택 사항)
df = pd.DataFrame(data, columns=['정상', '불량'], index=['주간근무', '야간근무'])
print("--- 입력된 분할표 데이터 ---")
print(df)
print("-" * 30)

# 2. 독립성 검정 수행
# correction=False는 Yates의 연속성 보정을 하지 않는 옵션입니다 (샘플 수가 충분할 때 사용).
# 만약 셀의 빈도가 5 미만으로 작다면 correction=True(기본값)를 쓰거나 피셔의 정확검정을 씁니다.
chi2, p, dof, expected = stats.chi2_contingency(data, correction=False)

# 3. 결과 출력
print("=== 이원 카이제곱 검정 (독립성 검정) 결과 ===")
print(f"카이제곱 통계량 (Chi-square statistic): {chi2:.4f}")
print(f"p-값 (p-value): {p:.4f}")
print(f"자유도 (Degrees of Freedom): {dof}")
print("기대 빈도(Expected Values):\n", expected)

# 4. 가설 검정 결론
alpha = 0.05
if p < alpha:
    print(f"\n결론: p-value가 {alpha}보다 작으므로 귀무가설을 기각합니다.")
    print("-> 근무 조와 제품 불량률 간에는 통계적으로 유의미한 연관성이 있습니다.")
else:
    print(f"\n결론: p-value가 {alpha}보다 크므로 귀무가설을 채택합니다.")
    print("-> 근무 조와 제품 불량률은 서로 독립적이라고 볼 수 있습니다.")
```

```python
import pandas as pd
from scipy import stats

# 가상의 로우 데이터 생성 (예: 성별에 따른 스마트폰 브랜드 선호도)
raw_data = {
    '성별': ['남', '여', '남', '남', '여', '여', '남', '여', '남', '여'] * 10,
    '선호브랜드': ['A사', 'B사', 'A사', 'B사', 'B사', 'A사', 'A사', 'B사', 'B사', 'A사'] * 10
}
df_raw = pd.DataFrame(raw_data)

# 1. crosstab을 이용해 데이터프레임을 카이제곱용 분할표로 자동 변환
crosstab_result = pd.crosstab(df_raw['성별'], df_raw['선호브랜드'])
print("--- pd.crosstab으로 변환된 분할표 ---")
print(crosstab_result)
print("-" * 30)

# 2. 카이제곱 검정 진행
chi2, p, dof, expected = stats.chi2_contingency(crosstab_result)

print(f"카이제곱 통계량: {chi2:.4f}")
print(f"p-값: {p:.4f}")
```
