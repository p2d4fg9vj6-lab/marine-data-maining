```python
# 0-1. 라이브러리 로드 및 Google Drive 마운트
from google.colab import drive
from pathlib import Path
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from sklearn.metrics import mean_squared_error

# 드라이브 연결
drive.mount('/content/drive')

# 데이터 폴더 경로 설정 (본인의 파일 위치에 맞게 수정하세요)
# 예시: MyDrive/type3_week12
DATA_DIR = Path('/content/drive/MyDrive/type3_week12')
print('설정된 데이터 폴더:', DATA_DIR)

# 0-2. 파일 존재 여부 확인
required_files = ['retail_sales_reg1.csv', 'weather_reg2.csv']

print('\n[파일 존재 여부 확인]')
for fname in required_files:
    fpath = DATA_DIR / fname
    print(f'{fname}:', 'OK' if fpath.exists() else '없음')
```

```python
# 1-1. 데이터 불러오기
file_path1 = DATA_DIR / 'retail_sales_reg1.csv'
df1 = pd.read_csv(file_path1)

print('[데이터 상위 5행]')
print(df1.head())
print('\n[기본 정보]')
print(df1.info())
print('\n[기초통계]')
print(df1.describe())

# 1-2. 다중회귀모형 적합 (sales ~ ad_cost + staff + event_cnt)
model1 = smf.ols('sales ~ ad_cost + staff + event_cnt', data=df1).fit()

print('\n[회귀 요약 결과]')
print(model1.summary())

# 1-3. 핵심 결과만 따로 보기
print('\n[회귀계수]')
print(model1.params)

print('\n[p-value]')
print(model1.pvalues)

print('\n[R-squared]')
print(model1.rsquared)

print('\n[95% 신뢰구간]')
print(model1.conf_int())

# 1-4. 광고비 10단위 증가 효과 해석 및 유의 변수 추출 예시
beta_ad = model1.params['ad_cost']
change_ad_10 = beta_ad * 10

print('\n광고비 계수:', beta_ad)
print('광고비 10단위 증가 시 예상 매출 변화:', change_ad_10)

# 유의한 변수 목록 (p-value < 0.05)
sig_vars1 = model1.pvalues[model1.pvalues < 0.05].index.tolist()
print('\n유의한 변수:', sig_vars1)
```

```python
# 2-1. 데이터 불러오기
file_path2 = DATA_DIR / 'weather_reg2.csv'
df2 = pd.read_csv(file_path2)

print('[데이터 상위 5행]')
print(df2.head())
print('\n[기본 정보]')
print(df2.info())
print('\n[기초통계]')
print(df2.describe())

# 2-2. 다중회귀모형 적합 (temperature ~ solar + wind + o3)
model2 = smf.ols('temperature ~ solar + wind + o3', data=df2).fit()

print('\n[회귀 요약 결과]')
print(model2.summary())

# 2-3. 핵심 결과만 따로 보기
print('\n[회귀계수]')
print(model2.params)

print('\n[p-value]')
print(model2.pvalues)

print('\n[R-squared]')
print(model2.rsquared)

print('\n[95% 신뢰구간]')
print(model2.conf_int())

# 2-4. 특정 조건에서의 예측값 계산 (solar=100, wind=5, o3=30)
new_df = pd.DataFrame({
    'solar': [100],
    'wind': [5],
    'o3': [30]
})

pred = model2.predict(new_df)
print('\n[특정 조건 예측값]')
print(pred)

# 2-5. 예측 요약표 (평균 예측과 95% 신뢰구간 정보)
pred_frame = model2.get_prediction(new_df).summary_frame(alpha=0.05)
print('\n[예측 요약표]')
print(pred_frame)

# 2-6. 모델전체 오차지표(MSE / RMSE) 계산
y_true = df2['temperature']
y_pred = model2.predict(df2)

mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)

print('\n[오차지표]')
print('MSE :', mse)
print('RMSE:', rmse)
```
