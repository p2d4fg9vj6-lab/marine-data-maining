```python
import pandas as pd

# 데이터 로드
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch7/air_quality.csv")

# 1) IQR 계산
Q1 = df['CO2'].quantile(0.25)
Q3 = df['CO2'].quantile(0.75)
IQR = Q3 - Q1

# 2) 상한 및 하한 계산
upper = Q3 + 1.5 * IQR
lower = Q1 - 1.5 * IQR

# 3) 이상치 식별 및 개수 출력
outliers = df[(df['CO2'] < lower) | (df['CO2'] > upper)]
print(len(outliers)) # 결과: 304
```

```python
import pandas as pd

# 데이터 로드
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch8/drinks.csv")

# 1) 대륙별 맥주 소비량 평균 계산 및 최대 대륙 찾기
continent_avg = df.groupby("continent")['beer_servings'].mean()
top_continent = continent_avg.idxmax()

# 2) 해당 대륙 내 맥주 소비량 5번째 국가의 소비량 구하기
df_top = df[df['continent'] == top_continent]
df_top = df_top.sort_values('beer_servings', ascending=False)
result = df_top.iloc[4]['beer_servings']
print(result) # 결과: 313
```

```python
import pandas as pd

# 데이터 로드
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch8/tourist.csv")

# 1) 방문객 합계 및 관광객 비율 계산
df['방문객합계'] = df['관광'] + df['공무'] + df['사업'] + df['기타']
df['관광객비율'] = df['관광'] / df['방문객합계']

# 2) 조건에 맞는 값 찾기 (nlargest 활용 방법)
a = df.nlargest(2, '관광객비율').iloc[1]['사업']
b = df.nlargest(2, '관광').iloc[1]['공무']
print(a + b) # 결과: 441
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# 데이터 로드
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch8/chem.csv")

# 1) Min-Max 스케일링
scaler = MinMaxScaler()
df['co_scaled'] = scaler.fit_transform(df[['co']])
df['nmhc_scaled'] = scaler.fit_transform(df[['nmhc']])

# 2) 표준편차 계산 및 3) 차이 계산 (반올림)
co_std = df['co_scaled'].std()
nmhc_std = df['nmhc_scaled'].std()
std_diff = round(co_std - nmhc_std, 3)
print(std_diff) # 결과: -0.017
```

```python
import pandas as pd

# 데이터 로드
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/main/part4/ch9/loan.csv")

# 1) 총대출액 컬럼 생성
df['총대출액'] = df['신용대출'] + df['담보대출']

# 2) 지역코드와 성별별 총대출액 합계 (피벗 테이블 활용)
pivot_result = df.pivot_table(index='지역코드', columns='성별', values='총대출액', aggfunc='sum')

# 3) 성별 간 차이가 가장 큰 지역코드 찾기
pivot_result['차이'] = abs(pivot_result[1] - pivot_result[2])
result = pivot_result['차이'].idxmax()
print(result) # 결과: 4100000278
```

```python
import pandas as pd

# 데이터 로드
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/main/part4/ch9/crime.csv")

# 1) 발생건수와 검거건수 분리
df1 = df[df["구분"] == "발생건수"].iloc[:, 2:].reset_index(drop=True)
df2 = df[df["구분"] == "검거건수"].iloc[:, 2:].reset_index(drop=True)

# 2) 검거율 계산 및 3) 연도별 최고 검거율 범죄유형 찾기
df3 = df2 / df1
max_crime_types = df3.idxmax(axis=1)

# 4) 해당 범죄유형의 검거건수 합산
total_arrests = 0
for idx, crime_type in enumerate(max_crime_types):
    total_arrests += df2.loc[idx, crime_type]
print(total_arrests) # 결과: 7799
```

```python
import pandas as pd

# 데이터 로드
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert_v2/main/part4/ch9/hr.csv")

# 1) 만족도 결측치 처리 (전체 평균)
df["만족도"] = df["만족도"].fillna(df["만족도"].mean())

# 2) 근속연수 결측치 처리 (부서와 성과등급 기준 평균)
gm = df.groupby(["부서", "성과등급"])["근속연수"].transform("mean").astype(int)
df["근속연수"] = df["근속연수"].fillna(gm)

# 3) 연봉/근속연수 기준 3위의 근속연수 (A)
df["연봉_근속연수"] = df["연봉"] / df["근속연수"]
A = df.nlargest(3, '연봉_근속연수').iloc[-1]["근속연수"]

# 4) 연봉/만족도 기준 2위의 교육참가횟수 (B)
df["연봉_만족도"] = df["연봉"] / df["만족도"]
B = df.nlargest(2, '연봉_만족도').iloc[-1]["교육참가횟수"]

# 5) 결과 출력
print(A + B) # 결과: 7.0
```
