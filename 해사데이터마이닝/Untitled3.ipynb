```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

# 2. 라이브러리 및 데이터 불러오기 [cite: 232, 235, 236]
train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part2/ch6/diabetes_train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part2/ch6/diabetes_test.csv")

# 3. 탐색적 데이터 분석 (EDA) [cite: 251-271]
# 데이터 크기, 결측치, 타겟(Outcome) 분포 확인
print(train.shape, test.shape)
print(train.isnull().sum().sum())
print(train['Outcome'].value_counts())

# 4. 데이터 전처리 [cite: 282]
# 타겟 컬럼 분리 및 불필요한 전처리는 생략(결측치 없음)
target = train.pop('Outcome')

# 5. 검증 데이터 분할 [cite: 291, 292]
X_tr, X_val, y_tr, y_val = train_test_split(train, target, test_size=0.2, random_state=0)

# 6. 머신러닝 학습 및 평가 [cite: 298-301]
rf = RandomForestClassifier(random_state=0)
rf.fit(X_tr, y_tr)
# 예측값(확률) 추출 [cite: 301, 306]
pred_val = rf.predict_proba(X_val)[:, 1]

# 7. 예측 및 결과 파일 생성 [cite: 306-308]
pred_test = rf.predict_proba(test)[:, 1]
submit = pd.DataFrame({'pred': pred_test})
submit.to_csv("result.csv", index=False)

# 결과 확인 [cite: 311]
print(pd.read_csv("result.csv").head())
```

① 데이터 전처리 (#4. Data Preprocessing)가 필요한 이유

컴퓨터와의 의사소통: 머신러닝 모델은 숫자 데이터만 이해할 수 있습니다. 문자형(범주형) 데이터를 원-핫 인코딩 등을 통해 숫자로 변환해야 모델이 계산을 수행할 수 있습니다.

왜곡 방지: 데이터의 스케일이 서로 크게 다를 경우, 특정 변수가 모델에 과도한 영향을 줄 수 있습니다. 이를 조정하여 모델의 성능을 안정화합니다.

② 검증 데이터 분할 (#5. Validation Split)이 필요한 이유

실전 모의고사: 학습 데이터(Train)로만 평가하면 모델이 해당 데이터에만 과하게 익숙해지는 과적합(Overfitting) 문제가 발생합니다. 별도의 검증 데이터를 통해 모델의 진짜 실력을 미리 확인해야 합니다.데이터 누수(Data Leakage)

방지: 평가 데이터를 미리 학습에 사용하는 실수를 방지하여, 제출 후 실제 채점에서 점수가 급락하는 것을 막아줍니다.

③ 머신러닝 학습 및 평가 (#6. Learning & Evaluation)가 필요한 이유

패턴의 자동화: 수만 개의 데이터를 인간이 일일이 분석하는 대신, 알고리즘(예: 랜덤 포레스트)을 통해 변수 간의 복잡한 상관관계를 자동으로 학습하여 최적의 예측 규칙을 찾아냅니다.

성능 보장: 문제에서 요구하는 지표(예: ROC-AUC)를 기준으로 모델을 평가함으로써, 목표로 하는 합격 기준에 도달했는지 객관적으로 판단할 수 있게 합니다.

④ 결과 파일 생성 (#7. Prediction & Submission)이 필요한 이유

최종 목적 달성: 분석의 모든 과정은 결국 **'새로운 데이터(Test)에 대한 정답'**을 내놓기 위함입니다.

규격 준수: 아무리 성능이 좋아도 시험에서 요구하는 특정 파일명(result.csv)과 컬럼 형식(pred)을 맞추지 않으면 0점 처리되므로, 이를 정확히 구현하는 것이 필수적입니다.
