```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score

# 2. 라이브러리 및 데이터 불러오기
train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part2/ch7/score_train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part2/ch7/score_test.csv")

# 3. 탐색적 데이터 분석(EDA)
print("Train Shape:", train.shape)
# print(train.info()) # 상세 정보 확인 시 주석 해제

# 4. 데이터 전처리
target = train.pop('Credit_Score') # 타겟 컬럼 분리
train = pd.get_dummies(train)      # 원-핫 인코딩
test = pd.get_dummies(test)

# 학습 데이터와 테스트 데이터의 컬럼 순서 맞추기
test = test.reindex(columns = train.columns, fill_value=0)

# 5. 검증 데이터 분할
X_tr, X_val, y_tr, y_val = train_test_split(train, target, test_size=0.2, random_state=0)

# 6. 머신러닝 학습 및 평가 (RandomForest)
rf = RandomForestClassifier(random_state=0)
rf.fit(X_tr, y_tr)
pred_val = rf.predict(X_val)
print('f1-macro score:', f1_score(y_val, pred_val, average='macro'))

# 7. 예측 및 결과 파일 생성
pred_test = rf.predict(test)
submit = pd.DataFrame({'pred': pred_test})
submit.to_csv("result.csv", index=False)

print("파일 저장 완료: result.csv")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
import numpy as np

# 2. 라이브러리 및 데이터 불러오기
train = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch5/train.csv")
test = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part4/ch5/test.csv")

# 3. 탐색적 데이터 분석(EDA)
print("Train Shape:", train.shape)
print(train.head(3))

# 4. 데이터 전처리
# 회귀 문제에서는 타겟인 'price'를 분리합니다.
target = train.pop('price')
train = pd.get_dummies(train)
test = pd.get_dummies(test)

# 학습 데이터와 테스트 데이터의 컬럼 순서 맞추기
test = test.reindex(columns = train.columns, fill_value=0)

# 5. 검증 데이터 분할
X_tr, X_val, y_tr, y_val = train_test_split(train, target, test_size=0.2, random_state=0)

# 6. 머신러닝 학습 및 평가 (회귀용 RandomForestRegressor 사용)
rf = RandomForestRegressor(random_state=0)
rf.fit(X_tr, y_tr)
pred_val = rf.predict(X_val)

# RMSE 계산 (평균 제곱근 오차)
rmse = np.sqrt(mean_squared_error(y_val, pred_val))
print('Validation RMSE:', rmse)

# 7. 예측 및 결과 파일 생성
pred_test = rf.predict(test)
submit = pd.DataFrame({'pred': pred_test})
submit.to_csv("result_regression.csv", index=False)

print("회귀 결과 저장 완료: result_regression.csv")
```
